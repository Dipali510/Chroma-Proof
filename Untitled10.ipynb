```python
# ==============================================================================
# CHROMAPROOF (SIH26231) — COMPLETE MULTI-KIT APPLICATION FOR GOOGLE COLAB
# ==============================================================================
# Run this entire cell. It will:
# 1. Install dependencies
# 2. Ingest real-world sample images via KaggleHub (with automatic synthetic fallback)
# 3. Initialize the multi-kit CV engine, Ed25519 notary, SQLite DB, and PDF generator
# 4. Launch a public web link accessible from phones, tablets, and laptops
# ==============================================================================

!pip install -q gradio opencv-python-headless cryptography reportlab pillow numpy pandas kagglehub

import os
import io
import json
import uuid
import sqlite3
import hashlib
import glob
from dataclasses import dataclass
from typing import Dict, Tuple, List, Optional
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import cv2
import gradio as gr
import kagglehub

from cryptography.hazmat.primitives.asymmetric import ed25519
from cryptography.hazmat.primitives import serialization
from cryptography.exceptions import InvalidSignature

from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, HRFlowable, Image as RLImage
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors

# ------------------------------------------------------------------------------
# 1. SAMPLE DATASET INGESTION
# ------------------------------------------------------------------------------
print("==> Checking sample data...")
sample_images = []
try:
    kaggle_path = kagglehub.dataset_download("ultralytics/medical-pills")
    sample_images = glob.glob(f"{kaggle_path}/**/*.jpg", recursive=True)[:6]
    print(f"Loaded {len(sample_images)} real images from Kaggle.")
except Exception as e:
    print(f"KaggleHub fallback active: {e}")

# Generate clean synthetic test targets if Kaggle download fails
if not sample_images:
    os.makedirs("sample_images", exist_ok=True)
    # Target 1: Scott Reagent Positive (Strong Blue)
    blue_img = np.full((400, 500, 3), (180, 50, 30), dtype=np.uint8)
    cv2.putText(blue_img, "DEMO: Cocaine Positive", (30, 200), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
    path1 = "sample_images/cocaine_positive.jpg"
    cv2.imwrite(path1, blue_img)

    # Target 2: Marquis Reagent Positive (Dark Purple)
    purple_img = np.full((400, 500, 3), (90, 20, 70), dtype=np.uint8)
    cv2.putText(purple_img, "DEMO: Marquis Positive", (30, 200), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
    path2 = "sample_images/marquis_positive.jpg"
    cv2.imwrite(path2, purple_img)

    sample_images = [path1, path2]

# ------------------------------------------------------------------------------
# 2. ROLE 3: CRYPTOGRAPHIC ENGINE & CANONICAL SERIALIZATION
# ------------------------------------------------------------------------------
class ChromaProofCrypto:
    def __init__(self):
        self._private_key = ed25519.Ed25519PrivateKey.generate()
        self._public_key = self._private_key.public_key()

    @property
    def public_key_hex(self) -> str:
        return self._public_key.public_bytes(
            encoding=serialization.Encoding.Raw,
            format=serialization.PublicFormat.Raw
        ).hex()

    @property
    def key_id(self) -> str:
        return f"KEY-ED25519-{self.public_key_hex[:8].upper()}"

    @staticmethod
    def sha256_digest(data: bytes) -> str:
        return hashlib.sha256(data).hexdigest()

    @staticmethod
    def canonicalize(record_dict: dict) -> str:
        subset = {
            "record_id": str(record_dict["record_id"]),
            "operator_id": str(record_dict["operator_id"]),
            "timestamp_utc": str(record_dict["timestamp_utc"]),
            "kit_id": str(record_dict["kit_id"]),
            "result": str(record_dict["result"]),
            "confidence": str(record_dict["confidence"]),
            "classifier_version": str(record_dict["classifier_version"]),
            "image_sha256": str(record_dict["image_sha256"]),
            "location": str(record_dict.get("location", "N/A"))
        }
        return json.dumps(subset, sort_keys=True, separators=(',', ':'), ensure_ascii=False)

    def sign(self, record_dict: dict) -> str:
        canonical_bytes = self.canonicalize(record_dict).encode("utf-8")
        return self._private_key.sign(canonical_bytes).hex()

    @classmethod
    def verify(
        cls,
        record_dict: dict,
        signature_hex: str,
        public_key_hex: str,
        image_bytes: bytes = None
    ) -> Tuple[bool, str]:
        if image_bytes is not None:
            calc_hash = cls.sha256_digest(image_bytes)
            if calc_hash != record_dict.get("image_sha256"):
                return False, f"IMAGE TAMPER DETECTED: Computed ({calc_hash[:10]}...) != Stored ({record_dict.get('image_sha256')[:10]}...)"

        try:
            pub_key = ed25519.Ed25519PublicKey.from_public_bytes(bytes.fromhex(public_key_hex))
            canonical_bytes = cls.canonicalize(record_dict).encode("utf-8")
            pub_key.verify(bytes.fromhex(signature_hex), canonical_bytes)
            return True, "VERIFIED: Cryptographic Ed25519 signature valid. Record payload is untampered."
        except InvalidSignature:
            return False, "TAMPER DETECTED: Digital signature mismatch. Stored record contents were altered."
        except Exception as ex:
            return False, f"VERIFICATION ERROR: {str(ex)}"

# ------------------------------------------------------------------------------
# 3. ROLE 1: MULTI-KIT COMPUTER VISION PIPELINE
# ------------------------------------------------------------------------------
@dataclass
class KitProfile:
    kit_id: str
    display_name: str
    target_substance: str
    pos_centroid_lab: Tuple[float, float, float]
    neg_centroid_lab: Tuple[float, float, float]
    pos_threshold_delta_e: float
    neg_threshold_delta_e: float
    description: str

class MultiKitRegistry:
    KITS: Dict[str, KitProfile] = {
        "cobalt_thiocyanate": KitProfile(
            kit_id="cobalt_thiocyanate",
            display_name="Cobalt Thiocyanate (Scott Reagent)",
            target_substance="Cocaine HCl / Base",
            pos_centroid_lab=(36.0, 3.0, -48.0),   # Distinct Blue
            neg_centroid_lab=(55.0, 45.0, 12.0),   # Pinkish-Red unreacted
            pos_threshold_delta_e=32.0,
            neg_threshold_delta_e=28.0,
            description="Presumptive test for cocaine salt and base."
        ),
        "marquis": KitProfile(
            kit_id="marquis",
            display_name="Marquis Reagent",
            target_substance="Opiates / MDMA / Amphetamines",
            pos_centroid_lab=(21.0, 28.0, -18.0),  # Dark Purple / Violet
            neg_centroid_lab=(82.0, 0.0, 8.0),     # Pale clear / straw
            pos_threshold_delta_e=30.0,
            neg_threshold_delta_e=24.0,
            description="Presumptive test for alkaloids, MDMA, and opiates."
        ),
        "duquenois_levine": KitProfile(
            kit_id="duquenois_levine",
            display_name="Duquenois-Levine Reagent",
            target_substance="Cannabinoids (THC)",
            pos_centroid_lab=(30.0, 31.0, -30.0),  # Indigo / Deep Violet
            neg_centroid_lab=(75.0, 2.0, 22.0),    # Amber / Clear
            pos_threshold_delta_e=28.0,
            neg_threshold_delta_e=24.0,
            description="Presumptive test for cannabis and hashish resin."
        )
    }

class ChromaProofVision:
    VERSION = "ChromaProof-CV-v2.1-Colab"

    @classmethod
    def evaluate(cls, bgr_image: np.ndarray, kit_id: str) -> dict:
        gray = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2GRAY)
        h, w = gray.shape

        # 1. Quality Gate: Blur & Glare
        lap_var = float(cv2.Laplacian(gray, cv2.CV_64F).var())
        glare_ratio = float(np.sum(gray > 248) / gray.size)

        if lap_var < 35.0:
            return {
                "quality_status": "FAILED_QUALITY_GATE",
                "quality_reasons": [f"Image blurred (Laplacian: {lap_var:.1f}, min: 35.0)"],
                "result": "INCONCLUSIVE",
                "confidence": "LOW (0.0%)",
                "explanation": "Image sharpness check failed. Retake holding device steady.",
                "classifier_version": cls.VERSION,
                "metrics": {"blur_score": round(lap_var, 2), "glare_ratio": round(glare_ratio, 3)}
            }

        # 2. Gray-World White Balance Calibration
        img_float = bgr_image.astype(np.float32)
        mean_b = np.mean(img_float[:, :, 0])
        mean_g = np.mean(img_float[:, :, 1])
        mean_r = np.mean(img_float[:, :, 2])
        mean_gray = (mean_b + mean_g + mean_r) / 3.0 + 1e-6

        img_float[:, :, 0] *= (mean_gray / (mean_b + 1e-6))
        img_float[:, :, 1] *= (mean_gray / (mean_g + 1e-6))
        img_float[:, :, 2] *= (mean_gray / (mean_r + 1e-6))
        calibrated = np.clip(img_float, 0, 255).astype(np.uint8)

        # 3. Center Reagent ROI Extraction
        cy, cx = h // 2, w // 2
        dy, dx = int(h * 0.15), int(w * 0.15)
        roi = calibrated[cy - dy : cy + dy, cx - dx : cx + dx]

        # 4. Colorimetric Distance in CIE-L*a*b*
        roi_lab = cv2.cvtColor(roi, cv2.COLOR_BGR2LAB).astype(np.float32)
        mean_l = float(np.mean(roi_lab[:, :, 0]) * (100.0 / 255.0))
        mean_a = float(np.mean(roi_lab[:, :, 1]) - 128.0)
        mean_b = float(np.mean(roi_lab[:, :, 2]) - 128.0)
        measured_lab = np.array([mean_l, mean_a, mean_b])

        profile = MultiKitRegistry.KITS.get(kit_id, MultiKitRegistry.KITS["cobalt_thiocyanate"])
        delta_e_pos = float(np.linalg.norm(measured_lab - np.array(profile.pos_centroid_lab)))
        delta_e_neg = float(np.linalg.norm(measured_lab - np.array(profile.neg_centroid_lab)))

        metrics = {
            "blur_score": round(lap_var, 2),
            "glare_ratio": round(glare_ratio, 3),
            "delta_e_pos": round(delta_e_pos, 2),
            "delta_e_neg": round(delta_e_neg, 2)
        }

        # 5. Classification Outcome
        if delta_e_pos <= profile.pos_threshold_delta_e and delta_e_pos < delta_e_neg:
            conf = max(55.0, min(95.0, 100.0 - (delta_e_pos * 1.4)))
            return {
                "quality_status": "PASSED",
                "quality_reasons": [],
                "result": "POSITIVE",
                "confidence": f"HIGH ({conf:.1f}%)",
                "explanation": f"Reaction profile matches presumptive target: {profile.target_substance}.",
                "classifier_version": cls.VERSION,
                "metrics": metrics
            }
        elif delta_e_neg <= profile.neg_threshold_delta_e and delta_e_neg < delta_e_pos:
            conf = max(50.0, min(92.0, 100.0 - (delta_e_neg * 1.4)))
            return {
                "quality_status": "PASSED",
                "quality_reasons": [],
                "result": "NEGATIVE",
                "confidence": f"HIGH ({conf:.1f}%)",
                "explanation": "No characteristic target shift detected. Reagent baseline preserved.",
                "classifier_version": cls.VERSION,
                "metrics": metrics
            }
        else:
            return {
                "quality_status": "PASSED",
                "quality_reasons": [],
                "result": "INCONCLUSIVE",
                "confidence": "UNCERTAIN (45.0%)",
                "explanation": "Ambiguous chromatic profile. Confirmatory laboratory testing required.",
                "classifier_version": cls.VERSION,
                "metrics": metrics
            }

# ------------------------------------------------------------------------------
# 4. DATABASE & PDF GENERATOR
# ------------------------------------------------------------------------------
class ChromaProofDB:
    def __init__(self, db_path="chromaproof_colab.db"):
        self.db_path = db_path
        self._init_db()

    def _init_db(self):
        with sqlite3.connect(self.db_path) as conn:
            conn.execute("""
                CREATE TABLE IF NOT EXISTS records (
                    record_id TEXT PRIMARY KEY,
                    operator_id TEXT NOT NULL,
                    timestamp_utc TEXT NOT NULL,
                    kit_id TEXT NOT NULL,
                    result TEXT NOT NULL,
                    confidence TEXT NOT NULL,
                    classifier_version TEXT NOT NULL,
                    explanation TEXT NOT NULL,
                    image_sha256 TEXT NOT NULL,
                    image_bytes BLOB NOT NULL,
                    canonical_payload TEXT NOT NULL,
                    signature_hex TEXT NOT NULL,
                    key_id TEXT NOT NULL,
                    public_key_hex TEXT NOT NULL,
                    sync_status TEXT NOT NULL,
                    location TEXT NOT NULL
                )
            """)
            conn.commit()

    def save(self, record: dict, img_bytes: bytes):
        with sqlite3.connect(self.db_path) as conn:
            conn.execute("""
                INSERT INTO records VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
            """, (
                record["record_id"], record["operator_id"], record["timestamp_utc"],
                record["kit_id"], record["result"], record["confidence"],
                record["classifier_version"], record["explanation"], record["image_sha256"],
                img_bytes, record["canonical_payload"], record["signature_hex"],
                record["key_id"], record["public_key_hex"], record["sync_status"], record["location"]
            ))
            conn.commit()

    def query(self, result_filter: str = "ALL", search: str = "") -> List[dict]:
        with sqlite3.connect(self.db_path) as conn:
            conn.row_factory = sqlite3.Row
            q = "SELECT record_id, timestamp_utc, operator_id, kit_id, result, confidence, sync_status FROM records WHERE 1=1"
            params = []
            if result_filter != "ALL":
                q += " AND result = ?"
                params.append(result_filter)
            if search.strip():
                q += " AND (record_id LIKE ? OR operator_id LIKE ? OR kit_id LIKE ?)"
                t = f"%{search.strip()}%"
                params.extend([t, t, t])
            q += " ORDER BY timestamp_utc DESC"
            cursor = conn.cursor()
            cursor.execute(q, params)
            return [dict(r) for r in cursor.fetchall()]

    def get(self, record_id: str) -> Optional[dict]:
        with sqlite3.connect(self.db_path) as conn:
            conn.row_factory = sqlite3.Row
            cursor = conn.cursor()
            cursor.execute("SELECT * FROM records WHERE record_id = ?", (record_id,))
            r = cursor.fetchone()
            return dict(r) if r else None

class PDFEngine:
    @staticmethod
    def generate(record: dict, image_bytes: bytes) -> str:
        pdf_path = f"ChromaProof_Report_{record['record_id']}.pdf"
        doc = SimpleDocTemplate(pdf_path, pagesize=letter, rightMargin=36, leftMargin=36, topMargin=36, bottomMargin=36)
        styles = getSampleStyleSheet()

        title_style = ParagraphStyle('Title', parent=styles['Heading1'], fontSize=16, textColor=colors.HexColor('#0F172A'))
        meta_style = ParagraphStyle('Meta', parent=styles['Normal'], fontSize=8, textColor=colors.HexColor('#64748B'))
        warning_style = ParagraphStyle('Warn', parent=styles['Normal'], fontSize=9, textColor=colors.HexColor('#B91C1C'), backColor=colors.HexColor('#FEE2E2'), borderPadding=6)
        cell_style = ParagraphStyle('Cell', parent=styles['Normal'], fontSize=8)
        cell_bold = ParagraphStyle('CellB', parent=styles['Normal'], fontSize=8, fontName='Helvetica-Bold')

        elements = [
            Paragraph("CHROMAPROOF — EVIDENCE CHAIN LOG & SCREENING RECORD", title_style),
            Paragraph(f"Record ID: {record['record_id']} | Generated: {record['timestamp_utc']}", meta_style),
            Spacer(1, 8),
            HRFlowable(width="100%", thickness=1.5, color=colors.HexColor('#2563EB'), spaceAfter=8),
            Paragraph("<b>MANDATORY LEGAL DISCLAIMER:</b> Presumptive field test screening only. Does not replace accredited gas chromatography-mass spectrometry (GC-MS) laboratory confirmation.", warning_style),
            Spacer(1, 10)
        ]

        try:
            rl_img = RLImage(io.BytesIO(image_bytes), width=180, height=135)
        except Exception:
            rl_img = Paragraph("Image render failed.", cell_style)

        table_data = [
            [Paragraph("Field Operator", cell_bold), Paragraph(record["operator_id"], cell_style), rl_img],
            [Paragraph("Kit Profile", cell_bold), Paragraph(record["kit_id"], cell_style), ""],
            [Paragraph("Presumptive Result", cell_bold), Paragraph(f"<b>{record['result']}</b>", cell_style), ""],
            [Paragraph("Confidence Score", cell_bold), Paragraph(record["confidence"], cell_style), ""],
            [Paragraph("Classifier Version", cell_bold), Paragraph(record["classifier_version"], cell_style), ""],
            [Paragraph("Field Location", cell_bold), Paragraph(record.get("location", "N/A"), cell_style), ""],
            [Paragraph("SHA-256 Digest", cell_bold), Paragraph(record["image_sha256"], cell_style), ""],
            [Paragraph("Key Identifier", cell_bold), Paragraph(record["key_id"], cell_style), ""],
            [Paragraph("Ed25519 Signature", cell_bold), Paragraph(record["signature_hex"][:32] + "...", cell_style), ""]
        ]

        t = Table(table_data, colWidths=[120, 240, 180])
        t.setStyle(TableStyle([
            ('SPAN', (2, 0), (2, -1)),
            ('VALIGN', (0, 0), (-1, -1), 'TOP'),
            ('GRID', (0, 0), (1, -1), 0.5, colors.HexColor('#CBD5E1')),
            ('BACKGROUND', (0, 0), (0, -1), colors.HexColor('#F8FAFC')),
            ('TOPPADDING', (0, 0), (-1, -1), 4),
            ('BOTTOMPADDING', (0, 0), (-1, -1), 4),
        ]))
        elements.append(t)
        doc.build(elements)
        return pdf_path

# Runtime Singletons
crypto_system = ChromaProofCrypto()
db_system = ChromaProofDB()

# ------------------------------------------------------------------------------
# 5. GRADIO CONTROLLER FUNCTIONS
# ------------------------------------------------------------------------------
def run_capture(input_image, operator_id, kit_type_display, location):
    if input_image is None:
        return "⚠️ Error: Please provide an image via camera or upload.", "{}", None, ""

    kit_key = "cobalt_thiocyanate"
    for k, profile in MultiKitRegistry.KITS.items():
        if profile.display_name == kit_type_display:
            kit_key = k
            break

    cv_bgr = cv2.cvtColor(np.array(input_image), cv2.COLOR_RGB2BGR)
    cv_res = ChromaProofVision.evaluate(cv_bgr, kit_key)

    _, buf = cv2.imencode(".jpg", cv_bgr)
    img_bytes = buf.tobytes()
    img_hash = crypto_system.sha256_digest(img_bytes)

    rec_id = f"CP-{datetime.now().strftime('%Y%m%d')}-{uuid.uuid4().hex[:6].upper()}"
    timestamp = datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")

    record_dict = {
        "record_id": rec_id,
        "operator_id": operator_id.strip() if operator_id.strip() else "OFFICER-4021",
        "timestamp_utc": timestamp,
        "kit_id": kit_key,
        "result": cv_res["result"],
        "confidence": cv_res["confidence"],
        "classifier_version": cv_res["classifier_version"],
        "explanation": cv_res["explanation"],
        "image_sha256": img_hash,
        "location": location.strip() if location.strip() else "GPS: Colab Dev Device"
    }

    canonical_str = crypto_system.canonicalize(record_dict)
    sig_hex = crypto_system.sign(record_dict)

    full_record = {
        **record_dict,
        "canonical_payload": canonical_str,
        "signature_hex": sig_hex,
        "key_id": crypto_system.key_id,
        "public_key_hex": crypto_system.public_key_hex,
        "sync_status": "SYNCED_LOCAL"
    }
    db_system.save(full_record, img_bytes)

    pdf_file = PDFEngine.generate(full_record, img_bytes)

    badge_color = "#DC2626" if cv_res["result"] == "POSITIVE" else "#16A34A" if cv_res["result"] == "NEGATIVE" else "#CA8A04"
    status_md = f"""
    <div style="padding: 12px; border-radius: 6px; background-color: {badge_color}22; border: 1px solid {badge_color};">
        <h2 style="color: {badge_color}; margin: 0;">Presumptive Outcome: {cv_res['result']}</h2>
        <p style="margin: 4px 0 0 0;"><strong>Kit:</strong> {kit_type_display} | <strong>Confidence:</strong> {cv_res['confidence']}</p>
        <p style="margin: 4px 0 0 0;"><strong>Record ID:</strong> <code>{rec_id}</code></p>
        <p style="margin: 4px 0 0 0;"><strong>Integrity Digest:</strong> <code>{img_hash[:16]}...</code></p>
    </div>
    """

    return status_md, json.dumps(full_record, indent=2), pdf_file, rec_id

def query_records_ui(filter_res, search_term):
    rows = db_system.query(filter_res, search_term)
    cols = ["record_id", "timestamp_utc", "operator_id", "kit_id", "result", "confidence", "sync_status"]
    if not rows:
        return pd.DataFrame(columns=cols), "No records found matching criteria."
    df = pd.DataFrame(rows)[cols]
    return df, f"Found {len(rows)} matching record(s)."

def verify_record_ui(record_id_input, simulate_tamper):
    if not record_id_input.strip():
        return "⚠️ Please enter a Record ID."

    rec = db_system.get(record_id_input.strip())
    if not rec:
        return f"❌ Record ID `{record_id_input}` does not exist in local database."

    test_dict = {
        "record_id": rec["record_id"],
        "operator_id": rec["operator_id"],
        "timestamp_utc": rec["timestamp_utc"],
        "kit_id": rec["kit_id"],
        "result": rec["result"],
        "confidence": rec["confidence"],
        "classifier_version": rec["classifier_version"],
        "image_sha256": rec["image_sha256"],
        "location": rec["location"]
    }

    if simulate_tamper:
        test_dict["result"] = "NEGATIVE" if test_dict["result"] == "POSITIVE" else "POSITIVE"

    valid, msg = ChromaProofCrypto.verify(
        test_dict,
        rec["signature_hex"],
        rec["public_key_hex"],
        rec["image_bytes"]
    )

    verdict_badge = "✅ VALID & AUTHENTIC" if valid else "❌ INTEGRITY COMPROMISED / TAMPER DETECTED"
    return f"## {verdict_badge}\n\n- **Status:** {msg}\n- **Record ID:** `{rec['record_id']}`\n- **Signer Key:** `{rec['key_id']}`"

# ------------------------------------------------------------------------------
# 6. GRADIO FRONTEND INTERFACE
# ------------------------------------------------------------------------------
kit_choices = [p.display_name for p in MultiKitRegistry.KITS.values()]

with gr.Blocks(title="ChromaProof Companion", theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🧪 ChromaProof: SIH26231 Multi-Kit Companion")
    gr.Markdown("> **PRESUMPTIVE SCREENING NOTICE:** Field colorimetric assays require accredited laboratory confirmation (GC/MS).")

    with gr.Tabs():
        with gr.TabItem("📸 Capture & Sealing"):
            with gr.Row():
                with gr.Column():
                    input_img = gr.Image(type="pil", label="Field Reagent Well Capture")

                    if sample_images:
                        gr.Examples(
                            examples=[[img] for img in sample_images],
                            inputs=input_img,
                            label="Sample Ingestion Files"
                        )

                    op_id = gr.Textbox(value="OFFICER-4021", label="Operator ID")
                    kit_select = gr.Dropdown(choices=kit_choices, value=kit_choices[0], label="Target Chemical Test Kit")
                    loc = gr.Textbox(value="GPS: 17.6868° N, 83.2185° E", label="Location Tag")
                    btn_capture = gr.Button("⚡ Analyze & Cryptographically Seal", variant="primary")

                with gr.Column():
                    status_view = gr.Markdown("*Awaiting sample image input...*")
                    json_view = gr.Code(language="json", label="RFC 8785 Sealed Evidence Record")
                    pdf_view = gr.File(label="Export Official PDF Report")
                    last_record_id = gr.Textbox(visible=False)

            btn_capture.click(
                fn=run_capture,
                inputs=[input_img, op_id, kit_select, loc],
                outputs=[status_view, json_view, pdf_view, last_record_id]
            )

        with gr.TabItem("📋 Chain of Custody History"):
            with gr.Row():
                filter_dd = gr.Dropdown(choices=["ALL", "POSITIVE", "NEGATIVE", "INCONCLUSIVE"], value="ALL", label="Filter Outcome")
                search_box = gr.Textbox(placeholder="Search ID, Operator, or Kit...", label="Query")
                btn_filter = gr.Button("🔍 Filter Log")

            history_grid = gr.Dataframe(label="Stored Evidence Chain")
            history_text = gr.Markdown()

            btn_filter.click(fn=query_records_ui, inputs=[filter_dd, search_box], outputs=[history_grid, history_text])
            demo.load(fn=query_records_ui, inputs=[filter_dd, search_box], outputs=[history_grid, history_text])

        with gr.TabItem("🔐 Verification & Anti-Tamper"):
            with gr.Row():
                with gr.Column():
                    rec_id_in = gr.Textbox(label="Record ID to Verify", placeholder="CP-YYYYMMDD-XXXXXX")
                    tamper_toggle = gr.Checkbox(label="Simulate Malicious Field Modification", value=False)
                    btn_verify = gr.Button("🔒 Verify Ed25519 Signature", variant="primary")
                with gr.Column():
                    verify_out = gr.Markdown("*Enter a Record ID to evaluate integrity.*")

            btn_verify.click(fn=verify_record_ui, inputs=[rec_id_in, tamper_toggle], outputs=verify_out)

# Launching with public share link allows browser access from mobile phones/tablets
demo.launch(share=True, debug=False)
```
